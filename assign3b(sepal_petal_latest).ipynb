```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

```

load dataset

```python
df=sns.load_dataset('iris')

df.head()
```

#### Display the frequency of each element in the column

```python
df['species'].value_counts()
```

Step4: scatter plot using scatterplot()

```python
# Set the figure size (width, height in inches) and resolution (DPI)
plt.figure(figsize=(10, 6), dpi=150)

# 's' controls the point marker size (default is usually ~20 to 50)

sns.scatterplot(data=df, x='petal_length', y='petal_width', s=10)

plt.xlabel('Petal Length', fontsize=12)
plt.ylabel('Petal Width', fontsize=12)
plt.title('Petal Length vs. Petal Width', fontsize=14)

plt.show()
```

### Line PLot (Seaborn)

```python
plt.figure(figsize=(15,5), dpi=150)
sns.lineplot(data=df, x='petal_length', y='petal_width', hue='species')
plt.xlabel('Petal Length', fontsize=12)
plt.ylabel('Petal Width', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(15,5), dpi=150)
sns.lineplot(data=df, x=df.index, y='sepal_length')
plt.xlabel('FLower ID')
plt.ylabel('sepal length')
plt.show()
```

```python
plt.figure(figsize=(10,5), dpi=150)
plt.xlabel("Petal Length")
plt.ylabel("Count")
sns.histplot(df['petal_length'], binwidth=0.25)
plt.title("HAHA my plot")
plt.show()
```
